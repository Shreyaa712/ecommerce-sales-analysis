```python
import pandas as pd

df = pd.read_csv("../data/processed/online_retail_cleaned.csv")

print("Dataset loaded successfully!")
```

```python
print("Shape:", df.shape)

df.head()
```

```python
products = df[["StockCode", "Description"]].drop_duplicates()

products.head()
```

```python
sales = df[["InvoiceNo", "StockCode", "Quantity", "UnitPrice", "TotalSales"]]

sales.head()
```

```python
merged_inner = pd.merge(
    sales,
    products,
    on="StockCode",
    how="inner"
)

merged_inner.head()
```

```python
print("Original sales shape:", sales.shape)
print("Merged shape:", merged_inner.shape)
print("Missing descriptions:", merged_inner["Description"].isnull().sum())
```
